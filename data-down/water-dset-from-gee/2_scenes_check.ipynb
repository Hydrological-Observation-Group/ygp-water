### check the selected landsat images


```python
import ee
import geemap
import geopandas as gpd
# ee.Authenticate(auth_mode='localhost')
ee.Initialize(project='earth-engine-auth-project')   
```

### parameters

```python
## true color composite bands
bands_vis_l57 = ['SR_B3', 'SR_B2', 'SR_B1']       # landsat 5,7
bands_vis_l89 = ['SR_B4', 'SR_B3', 'SR_B2']   # landsat 8,9
bands_vis_s2 = ['B4','B3','B2']  
## selected Bands (Landsat 5/7 SR)  
bands_sel_l57 = ['SR_B1', 'SR_B2', 'SR_B3', 'SR_B4', 'SR_B5', 'SR_B7']  
bands_sel_l89 = ['SR_B2', 'SR_B3', 'SR_B4', 'SR_B5', 'SR_B6', 'SR_B7']  
bands_sel_s2 = ['B2','B3','B4','B8','B11','B12']  

```

```python
path_ygp_vec = 'data/boundary/ygp_region.gpkg'  
ygp_gdf = gpd.read_file(path_ygp_vec) 
ygp_region = geemap.gdf_to_ee(ygp_gdf)   
```

### list all selected scenes 

```python
## landsat 5
## scene 01
l5_region_01 = ee.Geometry.Rectangle([106.06, 26.18, 106.71, 26.74])
l5_scene_01 = ee.Image('LANDSAT/LT05/C02/T1_L2/LT05_127042_19931227').clip(l5_region_01)\
                                .select(bands_sel_l57).multiply(0.0000275).add(-0.2).multiply(10000).toInt16() 

## landsat 7
## scene 01
l7_region_01 = ee.Geometry.Rectangle([99.95, 24.5, 100.62, 25.04])
l7_scene_01 = ee.Image('LANDSAT/LE07/C02/T1_L2/LE07_131043_20021208').clip(l7_region_01)\
                                .select(bands_sel_l57).multiply(0.0000275).add(-0.2).multiply(10000).toInt16() 

## landsat 8
# # scene 01
l8_region_01 = ee.Geometry.Rectangle([102.57, 24.8, 103.23, 25.40])
l8_scene_01 = ee.Image('LANDSAT/LC08/C02/T1_L2/LC08_129043_20140202').clip(l8_region_01)\
                                .select(bands_sel_l89).multiply(0.0000275).add(-0.2).multiply(10000).toInt16() 

## landsat 9
# # scene 01
l9_region_01 = ee.Geometry.Rectangle([107.88, 28.52, 108.53, 29.07])
l9_scene_01 = ee.Image('LANDSAT/LC09/C02/T1_L2/LC09_126040_20241030').clip(l9_region_01)\
                                .select(bands_sel_l89).multiply(0.0000275).add(-0.2).multiply(10000).toInt16()  

## sentinel-2
## sence 01
s2_region_01 = ee.Geometry.Rectangle([102.8, 26.81, 103.03, 27.0])
s2_scene_01 = ee.Image('COPERNICUS/S2_SR_HARMONIZED/20190407T033541_20190407T034651_T48RTQ').clip(s2_region_01).select(bands_sel_s2)

```

### show the selected scenes

```python
empty = ee.Image().byte()
Map = geemap.Map()
Map.addLayer(ygp_region, {}, 'Yunnan-Guizhou Plateau', True, 0.2)
Map.centerObject(ygp_region, 6)

## landsat 5
# // scene 01
l5_scene_01_outline = empty.paint(featureCollection=l5_region_01, color=1, width=3)
Map.addLayer(l5_scene_01_outline, {'palette': "#FF0000"}, 'l5_scene_01_outline')
Map.addLayer(l5_scene_01, {'bands': bands_vis_l57, 'min': 0, 'max': 2000}, 'l5 scene 01')


## landsat 7
# // scene 01
l7_scene_01_outline = empty.paint(featureCollection=l7_region_01, color=1, width=3)
Map.addLayer(l7_scene_01_outline, {'palette': '#0000FF'}, 'l7_scene_01_outline')
Map.addLayer(l7_scene_01, {'bands': bands_vis_l57, 'min': 0, 'max': 2000}, 'l7 scene 01')

## landsat 8
# // scene 01
l8_scene_01_outline = empty.paint(featureCollection=l8_region_01, color=1, width=3)
Map.addLayer(l8_scene_01_outline, {'palette': "#0C5707"}, 'l8_scene_01_outline')
Map.addLayer(l8_scene_01, {'bands': bands_vis_l89, 'min': 0, 'max': 2000}, 'l8 scene 01')


## landsat 9
# // scene 01
l9_scene_01_outline = empty.paint(featureCollection=l9_region_01, color=1, width=3)
Map.addLayer(l9_scene_01_outline, {'palette': "#F8F808"}, 'l8_scene_01_outline')
Map.addLayer(l9_scene_01, {'bands': bands_vis_l89, 'min': 0, 'max': 2000}, 'l9 scene 01')


## sentinel-2
# // scene 01
s2_scene_01_outline = empty.paint(featureCollection=s2_region_01, color=1, width=3)
Map.addLayer(s2_scene_01_outline, {'palette': '#9932CC'}, 's2_scene_01_outline')
Map.addLayer(s2_scene_01, {'bands': bands_vis_s2, 'min': 0, 'max': 2000}, 's2 scene 01')
Map

```
